```python
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
sns.set_style('darkgrid')
```

```python
df = pd.read_csv(r"C:\Users\user\Desktop\Work\Pthon for Data Analytics\banking.csv")
```

```python
df.head()
```

```python
df.drop(['previous', 'emp_var_rate', 'cons_price_idx', 'euribor3m', "nr_employed"], axis =1, inplace=True)
```

```python
df= df.replace(r'^\s*$',np.nan,regex=True)
```

```python
sns.countplot(
    x='marital',
    data=df,
    hue='marital',
    palette='Set2',
    linewidth=3,
    edgecolor='black',
    legend=False
)

plt.show()
```

```python
sns.countplot(
    x='month',
    data=df,
    hue='month',
    palette='Set2',
    linewidth=3,
    edgecolor='black',
    legend=False
)
plt.show()
```

```python
plt.figure(figsize=(18,6))
plt.subplot(1,2,1)
sns.countplot(x=df['y'], hue = df['day_of_week'], palette='summer',linewidth=3,edgecolor="white")
plt.title("Day_of_Week")


plt.subplot(1,2,2)
sns.countplot(x=df['y'], hue = df['poutcome'], palette='hot', linewidth = 3 , edgecolor = 'white')
plt.title("Pout_come")
plt.show()
```

```python
sns.countplot(x=df['contact'], hue = df['contact'], palette ='coolwarm',linewidth=2,edgecolor="black")
plt.show()
```

```python
print(df.columns.tolist())
```

```python
plt.figure(figsize=(16, 8))

sns.pointplot(x=df.columns[0], y="job", data=df,   linestyles="--", capsize=.3,    color="blue")

plt.show()
```

```python
plt.figure(figsize=(18, 6))

# First Graph: Marital Status
plt.subplot(1, 2, 1)
sns.countplot(
    data=df,
    x='marital',
    hue='marital',
    palette='viridis',
    legend=False,
    linewidth=2,
    edgecolor='black'
)
plt.title("STATUS")

# Second Graph: Duration
plt.subplot(1, 2, 2)
sns.countplot(
    data=df,
    x='duration',
    hue='duration',
    palette='coolwarm',
    legend=False,
    linewidth=2,
    edgecolor='black'
)
plt.title("TIME")

plt.show()
```
