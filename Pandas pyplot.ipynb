```python
# Drop irrelevant columns 

cars = cars.drop(['YEAR', 'Size', 'Make', 'Model', 'RATING'], axis=1)
cars.head(5)
```

```python
# Total no of columns & rows
cars.shape
```

```python
# Rows containing duplicate data 
duplicate_row_cars = cars[cars.duplicated()]
print("No of duplicate rows: ", duplicate_row_cars.shape)
```

```python
# Count the row before removing the data 
cars.count()
```

```python
# Drop the duplicates
cars = cars.drop_duplicates()
cars.count()
```

```python
# find the null values 
print(cars.isnull().sum())
```

```python
# Drop the missing values
cars= cars.dropna()
cars.count()
```

```python
print(cars.isnull().sum())
```

```python
# Finding the outliers
sns.boxplot(x=cars['TIME (h)'])


```

```python
# Plotting a Histogram for number of car per brand
cars.Make.value_counts().nlargest(40).plot(
    kind='bar',
    figsize=(10,5)
)
plt.title("Number of cars by make")
plt.ylabel("Number of cars")
plt.xlabel("Make");

plt.show()
```

```python
# Find the relationships between the variables

numeric_cars = cars.select_dtypes(include='number')

c = numeric_cars.corr()

plt.figure(figsize=(20,10))

sns.heatmap(c, cmap="BrBG", annot=True)

plt.show()

c
```
